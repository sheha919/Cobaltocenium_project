```python
# 1. Read your XYZ file into a string
file_path = "D:/Git/Cobaltocenium_project/bde_2026/co_cc2_oh_pcm.xyz"  # Change this to your actual file name
with open(file_path, "r") as f:
    xyz_data = f.read()

# 2. Create the viewer and add the data
view = py3Dmol.view(width=400, height=400)
view.addModel(xyz_data, "xyz")  # Pass the string, not the filename

# 3. Set the style (Stick or Sphere) and show
view.setStyle({'stick': {}})
view.zoomTo()
view.show()
```
